# Análise de Desempenho Acadêmico

## Objetivo

Este projeto tem como objetivo analisar o desempenho de uma turma de alunos,
considerando notas, frequência e situação de aprovação.

A análise busca identificar padrões no desempenho dos alunos e investigar
a relação entre frequência e média.

Os dados utilizados são fictícios e foram criados exclusivamente para fins
de estudo e demonstração de análise de dados com Python.

## Tecnologias utilizadas

- Python
- Pandas
- Matplotlib


## Base de dados

Os dados utilizados neste projeto são fictícios e foram criados exclusivamente para fins de estudo e demonstração.

```python
import pandas as pd
import matplotlib.pyplot as plt

dados = {
    "Aluno": ["Ana", "Bruno", "Carla", "Daniel", "Eduarda",
              "Felipe", "Gabriela", "Henrique", "Isabela", "Lucas"],
    "Nota 1": [8.0, 6.0, 9.0, 5.0, 7.5,
               4.0, 9.5, 6.5, 8.5, 3.5],
    "Nota 2": [9.0, 7.0, 8.5, 6.0, 8.0,
               5.5, 9.0, 5.0, 7.5, 4.5],
    "Frequência": [95, 85, 98, 75, 92,
                   68, 96, 82, 90, 60]
}

df = pd.DataFrame(dados)

df["Média"] = (df["Nota 1"] + df["Nota 2"]) / 2

df["Aprovado"] = df["Média"] >= 6

df
```

## Principais indicadores

**Critério de aprovação:** para fins deste projeto, foi considerada aprovação a média igual ou superior a 6,0. Esse critério é fictício e foi definido exclusivamente para fins de estudo.

```python
total_alunos = len(df)
aprovados = df["Aprovado"].sum()
reprovados = total_alunos - aprovados
taxa_aprovacao = (aprovados / total_alunos) * 100
media_geral = df["Média"].mean()
maior_media = df["Média"].max()
menor_media = df["Média"].min()

print("===== PRINCIPAIS INDICADORES =====")
print(f"Total de alunos: {total_alunos}")
print(f"Alunos aprovados: {aprovados}")
print(f"Alunos reprovados: {reprovados}")
print(f"Taxa de aprovação: {taxa_aprovacao:.1f}%")
print(f"Média geral: {media_geral:.2f}")
print(f"Maior média: {maior_media:.2f}")
print(f"Menor média: {menor_media:.2f}")
```

## Desempenho dos alunos

A análise individual permite observar a distribuição das médias e identificar
os alunos que ficaram acima ou abaixo do critério de aprovação utilizado.

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Aluno"], df["Média"])
plt.axhline(y=6, linestyle="--", label="Nota mínima para aprovação")

plt.title("Desempenho dos alunos")
plt.xlabel("Aluno")
plt.ylabel("Média")
plt.xticks(rotation=45)
plt.legend()

plt.show()
```

## Situação da turma

A turma apresentou 60% de aprovação, considerando a regra fictícia
estabelecida para este projeto.

```python
plt.figure(figsize=(6, 4))

plt.bar(
    ["Aprovados", "Reprovados"],
    [aprovados, reprovados]
)

plt.title("Situação dos alunos")
plt.ylabel("Quantidade de alunos")

plt.show()
```

## Relação entre frequência e média

Nesta etapa, analisamos se existe uma relação entre a frequência dos alunos
e suas médias.

```python
correlacao = df["Frequência"].corr(df["Média"])

plt.figure(figsize=(7, 5))

plt.scatter(df["Frequência"], df["Média"])

plt.title("Relação entre Frequência e Média")
plt.xlabel("Frequência (%)")
plt.ylabel("Média")

plt.show()

print(f"Correlação entre frequência e média: {correlacao:.2f}")
```

## Análise dos alunos reprovados

A análise dos alunos reprovados permite verificar se a reprovação está
relacionada apenas à frequência ou também ao desempenho nas avaliações.

```python
df[df["Aprovado"] == False][["Aluno", "Média", "Frequência"]]
```

Entre os alunos reprovados, foram identificados casos de baixa frequência
e também casos em que a frequência foi igual ou superior a 75%.

Isso indica que, nesta base fictícia, o desempenho nas avaliações também
teve papel importante no resultado final.

## Conclusões

- A turma possui 10 alunos, sendo 60% aprovados e 40% reprovados, considerando o critério fictício de média igual ou superior a 6,0.
- A média geral foi 6,88, com maior média de 9,25 e menor média de 4,00.
- Foi observada uma forte correlação positiva entre frequência e média (r = 0,97). Nesta base fictícia, alunos com maior frequência tendem a apresentar melhor desempenho.
- Entre os alunos reprovados, existem casos de baixa frequência e também casos com frequência igual ou superior a 75%, indicando que, nesta base, a frequência não foi o único fator relacionado ao resultado.
- A correlação observada não permite concluir que maior frequência, por si só, cause melhores notas.
- Os dados são fictícios e foram criados exclusivamente para estudo e demonstração.
